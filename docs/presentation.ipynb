Dieses Notebook präsentiert die Ergebnisse des Projekts zur Anwendung von Convolutional Neural Networks (CNN) zur Bilderkennung von Autos.

## Aufgabe 1: Entwicklung von CNN-Modellen

### a) CNN mit Keras/TensorFlow

Für die erste Aufgabe wurde ein CNN mit Keras/TensorFlow entwickelt, um Autos im CIFAR-10 Datensatz zu erkennen.

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
import os

# Verzeichnisse
models_dir = '../models'
data_dir = '../data'
results_dir = '../results'

# Laden der Testdaten
x_test = np.load(os.path.join(data_dir, 'x_test.npy'))
y_test = np.load(os.path.join(data_dir, 'y_test.npy'))
y_test_binary = np.load(os.path.join(data_dir, 'y_test_binary.npy'))

# Laden des trainierten Modells
model = load_model(os.path.join(models_dir, 'car_detection_model.keras'))

# Evaluierung des Modells
test_loss, test_accuracy = model.evaluate(x_test, y_test_binary)
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"Test Loss: {test_loss:.4f}")
```

### Visualisierung der Trainingsergebnisse

```python
# Anzeigen des Trainingsverlaufs
plt.figure(figsize=(12, 4))
img = plt.imread(os.path.join(models_dir, 'training_history.png'))
plt.imshow(img)
plt.axis('off')
plt.show()
```

### Visualisierung der Konfusionsmatrix

```python
# Anzeigen der Konfusionsmatrix
plt.figure(figsize=(8, 6))
img = plt.imread(os.path.join(models_dir, 'confusion_matrix.png'))
plt.imshow(img)
plt.axis('off')
plt.show()
```

### Visualisierung einiger Vorhersagen

```python
# Anzeigen einiger Vorhersagen
plt.figure(figsize=(10, 10))
img = plt.imread(os.path.join(models_dir, 'predictions.png'))
plt.imshow(img)
plt.axis('off')
plt.show()
```

## Aufgabe 2: Automerkennung auf Bildern

### a) Erkennung von Autos auf drei gegebenen Bildern

```python
# Anzeigen der Ergebnisse der Automerkennung auf drei gegebenen Bildern
plt.figure(figsize=(15, 5))
for i in range(3):
    plt.subplot(1, 3, i+1)
    img = plt.imread(os.path.join(results_dir, f'test_image_{i+1}_result.jpg'))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"Bild {i+1}")
plt.tight_layout()
plt.show()
```

### b) Erkennung von Autos auf drei weiteren Bildern

```python
# Anzeigen der Ergebnisse der Automerkennung auf drei weiteren Bildern
plt.figure(figsize=(15, 5))
for i in range(3):
    plt.subplot(1, 3, i+1)
    img = plt.imread(os.path.join(results_dir, f'additional_image_{i+1}_result.jpg'))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"Zusätzliches Bild {i+1}")
plt.tight_layout()
plt.show()
```

## Bonus: Personenerkennung

### Erkennung von Personen und Autos auf Bildern

```python
# Anzeigen der Ergebnisse der Personen- und Automerkennung
plt.figure(figsize=(15, 5))
for i in range(3):
    plt.subplot(1, 3, i+1)
    img = plt.imread(os.path.join('../bonus', f'bonus_image_{i+1}_result.jpg'))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"Bonus-Bild {i+1}")
plt.tight_layout()
plt.show()
```
